```python
from sentence_transformers import SentenceTransformer, util
import pandas as pd

# Load model
model = SentenceTransformer("all-MiniLM-L6-v2")

# . Load training data
df_train = pd.read_csv("training_data.csv").dropna()

train_questions = df_train["question"].astype(str).tolist()
train_answers = df_train["answer"].astype(str).tolist()

# . Load test data
df_test = pd.read_csv("testing_data.csv").dropna()

test_questions = df_test.iloc[:, 0].tolist()
test_answers = df_test.iloc[:, 1].tolist()
# . Encode training questions (corpus)
corpus_embeddings = model.encode(train_questions, convert_to_tensor=True)


test_embeddings = model.encode(test_questions, convert_to_tensor=True)

correct_count = 0
threshold = 0.75

#  Evaluation loop
for i in range(len(test_questions)):
    query_emb = test_embeddings[i]

    hits = util.semantic_search(query_emb, corpus_embeddings, top_k=1)
    idx = hits[0][0]["corpus_id"]

    predicted = train_answers[idx]

    sim = util.cos_sim(
        model.encode(predicted, convert_to_tensor=True),
        model.encode(test_answers[i], convert_to_tensor=True)
    ).item()

    if sim > threshold:
        correct_count += 1
    else:
        print("Query:", test_questions[i])
        print("Expected:", test_answers[i])
        print("Predicted:", predicted)
        print("Similarity:", sim)
        print()


accuracy = (correct_count / len(test_questions)) * 100

print("-" * 30)
print(f"FINAL ACCURACY: {accuracy:.2f}%")
```
